This notebook explores the verified Indian legal precedent corpus, demonstrates statutory provision extraction, and establishes text normalization routines.

```python
# Step 1: Clone / Load Corpus
import json
import pandas as pd
import matplotlib.pyplot as plt

with open('../app/datasets/indian_judgments.json', 'r') as f:
    corpus = json.load(f)

df = pd.DataFrame(corpus)
print(f"Total Judgments Loaded: {len(df)}")
df[['caseTitle', 'court', 'citation', 'category', 'outcome']].head()
```

```python
# Step 2: Distribution of Legal Categories
df['category'].value_counts().plot(kind='barh', title='Judgments by Legal Domain', color='#1E3E62')
plt.xlabel('Count')
plt.show()
```

```python
# Step 3: Text Normalization and Entity Extraction Demo
import sys
sys.path.append('..')
from app.preprocessing.text_cleaner import clean_legal_text, extract_statutory_provisions

sample_case = df.iloc[0]['judgmentText']
print("Raw snippet:", sample_case[:200])
print("Cleaned snippet:", clean_legal_text(sample_case)[:200])
print("Extracted Provisions:", extract_statutory_provisions(sample_case))
```
